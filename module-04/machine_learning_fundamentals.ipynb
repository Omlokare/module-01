## 1. Understand the workflow

A basic supervised machine learning workflow is:

1. Choose a question and labelled data.
2. Separate input features (`X`) from the target labels (`y`).
3. Split data into training and test sets.
4. Fit a model using training data only.
5. Evaluate predictions on held-out test data.
6. Use the trained model to predict a new example.

The test set gives an estimate of how well the model handles examples it did not train on.

## 2. Load the dataset

Iris is included with Scikit-learn, so no download is needed. Each flower has four measured features and one of three species labels.

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
X = iris.data
y = iris.target

print("Feature columns:", list(X.columns))
print("Target names:", list(iris.target_names))
print("Dataset shape:", X.shape)
display(X.head())
display(y.value_counts().sort_index().rename(index=dict(enumerate(iris.target_names))))
```

## 3. Split into training and test data

The model will learn from the training set. We will reserve 25% of examples for a final test. Stratification keeps the species proportions similar in both sets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Training examples:", len(X_train))
print("Test examples:", len(X_test))
```

## 4. Train a first model

Logistic Regression is a common classification model. A pipeline scales the numeric features before fitting the classifier. The pipeline ensures the scaling values are learned from training data only.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000, random_state=42)
)
model.fit(X_train, y_train)

print("Model training complete.")
```

## 5. Test and understand the results

Accuracy is the fraction of test predictions that are correct. A classification report also gives precision, recall, and F1 score for each species.

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = model.predict(X_test)
print(f"Test accuracy: {accuracy_score(y_test, y_pred):.3f}")
print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
print("Confusion matrix (rows = actual, columns = predicted):")
print(confusion_matrix(y_test, y_pred))
```

### How to read these metrics

- **Accuracy:** share of all test examples classified correctly.
- **Precision:** of examples predicted as a species, how many truly belong to it?
- **Recall:** of examples truly belonging to a species, how many did the model find?
- **F1 score:** a combined measure of precision and recall.
- **Confusion matrix:** shows which actual species were predicted as each species.

A good score on this small practice dataset does not guarantee the model will work equally well on new data from a different source.

## 6. Predict a new example

Enter measurements in the same feature order shown below. These are example values in centimeters, not a real flower measurement.

```python
example_flower = pd.DataFrame([{
    "sepal length (cm)": 5.1,
    "sepal width (cm)": 3.5,
    "petal length (cm)": 1.4,
    "petal width (cm)": 0.2,
}])

predicted_class = model.predict(example_flower)[0]
probabilities = model.predict_proba(example_flower)[0]
print("Predicted species:", iris.target_names[predicted_class])
print("Class probabilities:")
for species, probability in zip(iris.target_names, probabilities):
    print(f"  {species}: {probability:.3f}")
```
