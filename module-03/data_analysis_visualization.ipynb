# Module 03: Data Analysis & Visualization

Practice loading a sample dataset, checking and cleaning it with Pandas and NumPy, summarizing the results, and plotting with Matplotlib.

## 1. Import the tools

Run this notebook in Google Colab. Pandas, NumPy, and Matplotlib are available in the standard Colab runtime.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
```

## 2. Create and load a sample dataset

This small made-up dataset contains monthly sales by product. It includes one missing sales value and a duplicate row so we can practice cleaning.

```python
data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "product": ["Notebook", "Pen", "Notebook", "Pen", "Notebook", "Pen", "Notebook", "Pen", "Notebook", "Pen", "Notebook", "Pen"],
    "units_sold": [120, 180, 135, 195, 150, 210, 165, 225, 175, 240, np.nan, 260],
    "unit_price": [4.0, 1.5, 4.0, 1.5, 4.0, 1.5, 4.0, 1.5, 4.0, 1.5, 4.0, 1.5],
}
df = pd.DataFrame(data)
df = pd.concat([df, df.iloc[[2]]], ignore_index=True)  # Add one duplicate for practice
df.to_csv("sample_sales.csv", index=False)

# Load the CSV file, just as you would with a dataset file.
df = pd.read_csv("sample_sales.csv")
df.head()
```

## 3. Explore the dataset

Check the first rows, column types, dimensions, and missing values before analysis.

```python
print("Shape (rows, columns):", df.shape)
print("\nColumn information:")
df.info()
print("\nMissing values by column:")
print(df.isna().sum())
print("\nNumeric summary:")
df.describe()
```

## 4. Clean the data and add a calculated column

Remove duplicate records, fill the missing unit count with the median, and calculate revenue. In a real project, choose how to handle missing values based on what the data means.

```python
clean_df = df.drop_duplicates().copy()
median_units = clean_df["units_sold"].median()
clean_df["units_sold"] = clean_df["units_sold"].fillna(median_units)
clean_df["units_sold"] = clean_df["units_sold"].astype(int)
clean_df["revenue"] = clean_df["units_sold"] * clean_df["unit_price"]

print("Rows before cleaning:", len(df))
print("Rows after removing duplicates:", len(clean_df))
print("Missing values after cleaning:")
print(clean_df.isna().sum())
clean_df.head()
```

## 5. Analyze with Pandas and NumPy

Group the data by product and calculate total units and revenue. NumPy can also provide basic numerical calculations.

```python
product_summary = clean_df.groupby("product")[["units_sold", "revenue"]].sum().sort_values("revenue", ascending=False)
display(product_summary)

revenues = clean_df["revenue"].to_numpy()
print("Average monthly revenue:", round(np.mean(revenues), 2))
print("Median monthly revenue:", round(np.median(revenues), 2))
print("Total revenue:", round(np.sum(revenues), 2))
```

## 6. Bar chart: revenue by product

A bar chart helps compare categories.

```python
product_summary["revenue"].plot(kind="bar", color=["#3973ac", "#f2a541"])
plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Line chart: monthly revenue

A line chart is useful for seeing change over time. The rows are already in calendar order.

```python
monthly_revenue = clean_df.groupby("month", sort=False)["revenue"].sum()
monthly_revenue.plot(kind="line", marker="o", color="#3973ac")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 8. Pie chart: share of units sold

A pie chart shows each product's share of total units. For many categories, a bar chart is usually easier to compare.

```python
units_by_product = clean_df.groupby("product")["units_sold"].sum()
plt.figure(figsize=(6, 6))
plt.pie(units_by_product, labels=units_by_product.index, autopct="%1.1f%%", startangle=90)
plt.title("Share of Units Sold by Product")
plt.tight_layout()
plt.show()
```
